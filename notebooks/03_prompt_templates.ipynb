# 3 · Prompt templates with structured example data

Summarize a synthetic fixture without inventing a credit score or approval rule. This retains the original template exercise while avoiding unsupported lending decisions. Configure the root `.env` before running Azure cells. Requests may incur charges. Stored outputs are cleared; no new live Azure results are supplied.

```python
from pathlib import Path
import sys
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "lab/config.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository.")
sys.path.insert(0, str(ROOT))
from lab.config import get_chat_model
```

```python
import json
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
example = json.loads((ROOT / "assets/customer_example.json").read_text())
prompt = ChatPromptTemplate.from_messages([
    ("system", "Summarize only the supplied practice data. Do not invent missing facts, calculate a credit-risk score, "
     "or make an approval decision. State that the currency is unspecified. Use at most 150 words."),
    ("human", "Profile: {customer_profile}\nTransactions: {transaction_history}\nOther facts: {external_factors}"),
])
print("Template inputs:", prompt.input_variables)
```

```python
chain = prompt | get_chat_model() | StrOutputParser()
result = chain.invoke({name: json.dumps(example[name]) for name in prompt.input_variables})
print(result)
```

The example file is a newly constructed practice fixture. The source archive's named financial profile is excluded from the public package. No predictive credit model or validated scoring formula is provided.
