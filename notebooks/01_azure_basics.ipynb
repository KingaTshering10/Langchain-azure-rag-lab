Use your Azure chat deployment through the `langchain-openai` integration. Configure the root `.env` before running Azure cells. Requests may incur charges. Stored outputs are cleared; no new live Azure results are supplied.

```python
from pathlib import Path
import sys
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "lab/config.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository.")
sys.path.insert(0, str(ROOT))
from lab.config import get_chat_model
```

```python
llm = get_chat_model()
response = llm.invoke("Write a one-line Monday motivation quote.")
print(type(response).__name__)
print(response.content)
```
