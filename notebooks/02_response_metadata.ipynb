# 2 · Inspect the response and usage metadata

Configure the root `.env` before running Azure cells. Requests may incur charges. Stored outputs are cleared; no new live Azure results are supplied.

```python
from pathlib import Path
import sys
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "lab/config.py").is_file()), None)
if ROOT is None:
    raise RuntimeError("Open this notebook from inside the repository.")
sys.path.insert(0, str(ROOT))
from lab.config import get_chat_model
```

```python
llm = get_chat_model()
response = llm.invoke("Tell me a short joke.")
print(response.content)
print(response.response_metadata)
```

```python
usage = response.usage_metadata or response.response_metadata.get("token_usage", {})
print("Usage metadata:", usage)
total_tokens = usage.get("total_tokens") if isinstance(usage, dict) else None
print("Total tokens:", total_tokens if total_tokens is not None else "Not returned by this deployment")
```

Metadata fields depend on the deployment and provider response. Use optional lookups rather than assuming every response contains the same nested keys. Token usage is not a cost calculation.
